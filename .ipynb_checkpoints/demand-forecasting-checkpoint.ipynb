```python
# Importing dependencies
import sklearn
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import torch
import captum # for explanation methods (IG & LRP)
import torch.nn
import torch.optim
```

```python
# Loading the data
data = pd.read_csv("DataCoSupplyChainDataset.csv", encoding='latin-1')
```

```python
# EDA analysis
print(data.info())
missing = data.isnull().sum()
print(missing[missing > 0])
```

```python
# Preprocessing + train/test split
# Remove columns with too many missing values
data = data.drop(columns=['Order Zipcode', 'Product Description', 'Customer Lname'])
# removed Customer Lname because very little correlation expected with target and harder to impute NaN here.

y = data['Sales']
X = data.drop('Sales', axis=1)

# Divide into train / test split
X_train, X_test, y_train, y_test = sklearn.model_selection.train_test_split(X, y, random_state=42)

# Impute missing values for the other columns with missing values
imp = SimpleImputer(strategy='median')
X_train[['Customer Zipcode']] = imp.fit_transform(X_train[['Customer Zipcode']])
X_test[['Customer Zipcode']] = imp.transform(X_test[['Customer Zipcode']])

# Scale data for time series forecasting
X_scaler = MinMaxScaler()
X_train = X_scaler.fit_transform(X_train)
X_test = X_scaler.transform(X_test)

y_scaler = MinMaxScaler()
y_test = y_scaler.fit_transform(y_train)
y_test = y_scaler.transform(y_test)
```

```python
# Preprocessing continued

def create_sequence(X, y, seq_len):
    X_seq, y_seq = [], []
    for i in range(X.shape[1] - seq_len):
        new_seq_X = X[i : i+seq_len]
        new_seq_y = y[i+seq_len]
        X_seq.append(new_seq_X)
        y_seq.append(new_seq_y)
    return np.array(X_seq), np.array(y_seq)
    
# Create sequences in data for time series forecasting

# Turn numpy arrays into torch tensors 
X_train = torch.Tensor(X_train).unsqueeze(-1)
X_test = torch.Tensor(X_test).unsqueeze(-1)
y_train = torch.Tensor(y_train)
y_test = torch.Tensor(y_test)
```
